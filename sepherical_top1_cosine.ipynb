# Top-1 Weight-Norm Clustered FedAvg with Non-IID (Dirichlet) data

This notebook keeps the FedAvg training, checkpoint, resume, and metrics flow. After local training, client updates are clustered with elbow Spherical K-Means. Each cluster selects the client whose flattened local model-weight vector has the largest L2 norm, and the server performs sample-weighted FedAvg over the selected representatives.

```python
def seed_everything(seed):
    random.seed(seed)
    np.random.seed(seed % (2**32 - 1))
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

def _load_cifar10_from_drive(data_dir):
    batch_dir = data_dir / 'cifar-10-batches-py'
    if not batch_dir.is_dir():
        batch_dir = data_dir / 'batches-py'
    if not batch_dir.is_dir():
        raise FileNotFoundError(f'CIFAR-10 batches-py không tồn tại: {data_dir}')

    def read_batch(path):
        with path.open('rb') as handle:
            batch = pickle.load(handle, encoding='bytes')
        images = torch.tensor(batch[b'data']).reshape(-1, 3, 32, 32).float().div_(255.0)
        labels = torch.tensor(batch.get(b'labels', batch.get('labels')), dtype=torch.long)
        return images, labels

    train_parts = [read_batch(batch_dir / f'data_batch_{i}') for i in range(1, 6)]
    test_images, test_labels = read_batch(batch_dir / 'test_batch')
    train_images = torch.cat([part[0] for part in train_parts])
    train_labels = torch.cat([part[1] for part in train_parts])
    return train_images, train_labels, test_images, test_labels

if DATASET_NAME == 'mnist':
    train_ds = datasets.MNIST(DATA_DIR, train=True, download=False)
    test_ds = datasets.MNIST(DATA_DIR, train=False, download=False)
    train_images = train_ds.data.float().div_(255.0)
    test_images = test_ds.data.float().div_(255.0)
    train_labels = train_ds.targets.long()
    test_labels = test_ds.targets.long()
    DATA_MEAN, DATA_STD = 0.1307, 0.3081
    INPUT_CHANNELS, INPUT_SIZE = 1, 28
else:
    train_images, train_labels, test_images, test_labels = _load_cifar10_from_drive(DATA_DIR)
    DATA_MEAN = torch.tensor([0.4914, 0.4822, 0.4465]).view(1, 3, 1, 1)
    DATA_STD = torch.tensor([0.2023, 0.1994, 0.2010]).view(1, 3, 1, 1)
    INPUT_CHANNELS, INPUT_SIZE = 3, 32

IMAGENET_MEAN = torch.tensor([0.485, 0.456, 0.406], device=DEVICE).view(1, 3, 1, 1)
IMAGENET_STD = torch.tensor([0.229, 0.224, 0.225], device=DEVICE).view(1, 3, 1, 1)
if MODEL_NAME == 'cnn':
    train_features = ((train_images - DATA_MEAN) / DATA_STD)
    test_features = ((test_images - DATA_MEAN) / DATA_STD)
    if DATASET_NAME == 'mnist':
        train_features = train_features.unsqueeze(1)
        test_features = test_features.unsqueeze(1)
    train_features = train_features.to(DEVICE)
    test_features = test_features.to(DEVICE)
elif MODEL_NAME == 'logreg_raw':
    train_features = ((train_images - DATA_MEAN) / DATA_STD).reshape(len(train_images), -1).to(DEVICE)
    test_features = ((test_images - DATA_MEAN) / DATA_STD).reshape(len(test_images), -1).to(DEVICE)
else:
    train_features = None
    test_features = None
train_labels_gpu = train_labels.to(DEVICE)
test_labels_gpu = test_labels.to(DEVICE)
print('Train:', train_images.shape, 'Test:', test_images.shape)
```

```python
import re

def generate_dirichlet_partition(labels, num_clients=25, alpha=0.5, seed=42, min_samples=10, max_retries=100):
    labels_np = labels.cpu().numpy()
    num_samples = len(labels_np)
    classes = np.unique(labels_np)
    num_classes = len(classes)
    for attempt in range(max_retries):
        rng = np.random.default_rng(seed + attempt)
        client_sample_indices = [[] for _ in range(num_clients)]
        for c in classes:
            idx_c = np.where(labels_np == c)[0]
            rng.shuffle(idx_c)
            counts = rng.multinomial(len(idx_c), rng.dirichlet(np.repeat(alpha, num_clients)))
            start = 0
            for client_id, count in enumerate(counts):
                client_sample_indices[client_id].extend(idx_c[start:start + count])
                start += count
        if min(map(len, client_sample_indices)) >= min_samples:
            client_indices = []
            client_class_counts = torch.zeros((num_clients, num_classes), dtype=torch.long)
            for client_id, indices in enumerate(client_sample_indices):
                indices_arr = np.asarray(indices, dtype=np.int64)
                rng.shuffle(indices_arr)
                t_indices = torch.tensor(indices_arr, dtype=torch.long)
                client_indices.append(t_indices)
                client_class_counts[client_id] = torch.bincount(labels[t_indices], minlength=num_classes)
            assert sum(map(len, client_indices)) == num_samples
            assert len(torch.cat(client_indices).unique()) == num_samples
            return client_indices, client_class_counts
    raise RuntimeError(f'Could not generate Dirichlet partition with alpha={alpha}.')

# Parse PARTITION_TYPE exactly as in partition_dirichlet.ipynb.
def _parse_partition_type(partition_type):
    if partition_type == 'dirichlet':
        return 'dirichlet', None
    match = re.fullmatch(r'noniid_(\d+)', partition_type)
    if match:
        return 'noniid', int(match.group(1))
    raise ValueError(
        f"PARTITION_TYPE không hợp lệ: '{partition_type}'. "
        "Dùng 'dirichlet' hoặc 'noniid_N' (ví dụ: 'noniid_2')."
    )

partition_kind, classes_per_client = _parse_partition_type(PARTITION_TYPE)
if partition_kind == 'dirichlet':
    stem = f'{DATASET_NAME}_dirichlet_alpha{DIRICHLET_ALPHA}_{NUM_CLIENTS}_seed{PARTITION_SEED}'
else:
    stem = f'noniid_{classes_per_client}_{NUM_CLIENTS}clients_seed{PARTITION_SEED}'
if str(IMBALANCE_MODE).strip().lower() == 'global':
    stem += '_global'
elif str(IMBALANCE_MODE).strip().lower() != 'none':
    raise ValueError("IMBALANCE_MODE phải là 'None' hoặc 'Global'.")
partition_path = PARTITION_DIR / f'{stem}.pt'
if not partition_path.exists():
    raise FileNotFoundError(
        f'Partition không tồn tại: {partition_path}. '
        f'Hãy chạy partition_dirichlet.ipynb với PARTITION_TYPE=\"{PARTITION_TYPE}\" trước.'
    )
print(f'Loading {PARTITION_TYPE} partition from {partition_path}...')
partition = torch.load(partition_path, map_location='cpu', weights_only=False)
if partition.get('num_clients', NUM_CLIENTS) != NUM_CLIENTS:
    raise ValueError(
        f'Partition có num_clients={partition.get("num_clients")}, '
        f'nhưng notebook yêu cầu NUM_CLIENTS={NUM_CLIENTS}.'
    )
if partition.get('distribution') not in (None, PARTITION_TYPE):
    raise ValueError(
        f'Partition có distribution={partition.get("distribution")!r}, '
        f'không khớp PARTITION_TYPE={PARTITION_TYPE!r}.'
    )
client_indices = partition['client_indices']
assert len(client_indices) == NUM_CLIENTS
if 'class_counts' in partition:
    client_class_counts = partition['class_counts']
else:
    client_class_counts = torch.zeros(NUM_CLIENTS, 10, dtype=torch.long)
    for client_id, indices in enumerate(client_indices):
        client_class_counts[client_id] = torch.bincount(train_labels[indices], minlength=10)

partition_num_samples = sum(len(x) for x in client_indices)
assert len(torch.cat(client_indices).unique()) == partition_num_samples
client_sizes = np.array([len(x) for x in client_indices])
print(f'Client sizes: min={client_sizes.min()}, max={client_sizes.max()}, mean={client_sizes.mean():.1f}, std={client_sizes.std():.1f}')

plt.figure(figsize=(14, 6))
bottom = np.zeros(NUM_CLIENTS)
for cls, color in enumerate(plt.cm.tab10(np.linspace(0, 1, 10))):
    values = client_class_counts[:, cls].cpu().numpy()
    plt.bar(np.arange(NUM_CLIENTS), values, bottom=bottom, label=f'Class {cls}', color=color, edgecolor='black', linewidth=0.5)
    bottom += values
plt.xlabel('Client ID')
plt.ylabel('Number of samples')
plt.title(f'Non-IID {PARTITION_TYPE} Class Distribution (K={NUM_CLIENTS})')
plt.xticks(np.arange(NUM_CLIENTS))
plt.legend(ncol=10, loc='upper center', bbox_to_anchor=(0.5, -0.15))
plt.tight_layout()
partition_chart = PARTITION_DIR / f'{stem}_dist.png'
plt.savefig(partition_chart, dpi=150, bbox_inches='tight')
plt.show()
```

```python
class SmallCNN(nn.Module):
    def __init__(self, in_channels=1, input_size=28):
        super().__init__()
        self.features = nn.Sequential(nn.Conv2d(in_channels, 16, 5), nn.ReLU(), nn.MaxPool2d(2), nn.Conv2d(16, 32, 5), nn.ReLU(), nn.MaxPool2d(2))
        feature_size = (input_size - 5 + 1) // 2
        feature_size = (feature_size - 5 + 1) // 2
        self.classifier = nn.Sequential(nn.Flatten(), nn.Linear(32 * feature_size * feature_size, 128), nn.ReLU(), nn.Linear(128, 10))
    def forward(self, x):
        return self.classifier(self.features(x))

class LinearProbe(nn.Module):
    def __init__(self, embedding_dim=512):
        super().__init__()
        self.linear = nn.Linear(embedding_dim, 10)
    def forward(self, x):
        return self.linear(x)

class RawLogisticRegression(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.linear = nn.Linear(input_dim, 10)
    def forward(self, x):
        return self.linear(x)

def embedding_cache_filename(split):
    key = f'{DATASET_NAME}_resnet18_imagenet1k_v1_resize{RESNET_SIZE}_imagenet_norm_{split}'
    return f'resnet18_sz{RESNET_SIZE}_{split}_{hashlib.sha256(key.encode()).hexdigest()[:8]}.pt'

def preprocess_resnet(images):
    x = images.unsqueeze(1).repeat(1, 3, 1, 1) if DATASET_NAME == 'mnist' else images
    x = F.interpolate(x, size=(RESNET_SIZE, RESNET_SIZE), mode='bilinear', align_corners=False)
    return (x - IMAGENET_MEAN) / IMAGENET_STD

def make_embeddings(images, backbone):
    backbone.eval()
    outputs = []
    with torch.no_grad():
        for start in range(0, len(images), EMBED_BATCH_SIZE):
            outputs.append(backbone(preprocess_resnet(images[start:start + EMBED_BATCH_SIZE].to(DEVICE))).float().cpu())
    return torch.cat(outputs)

if MODEL_NAME == 'resnet_linear':
    train_embed_path = DATA_DIR / embedding_cache_filename('train')
    test_embed_path = DATA_DIR / embedding_cache_filename('test')
    if not (train_embed_path.exists() and test_embed_path.exists()):
        raise FileNotFoundError(
            f'Thiếu file embedding cache:\n  - {train_embed_path}\n  - {test_embed_path}\n'
            'Hãy chạy partition_dirichlet.ipynb trước để tạo embedding.'
        )
    train_features = torch.load(train_embed_path, map_location='cpu')
    test_features = torch.load(test_embed_path, map_location='cpu')
    train_features = train_features.to(DEVICE)
    test_features = test_features.to(DEVICE)

def build_model():
    if MODEL_NAME == 'cnn':
        return SmallCNN(INPUT_CHANNELS, INPUT_SIZE).to(DEVICE)
    if MODEL_NAME == 'logreg_raw':
        return RawLogisticRegression(INPUT_CHANNELS * INPUT_SIZE * INPUT_SIZE).to(DEVICE)
    return LinearProbe(512).to(DEVICE)

def batch_seed(round_id, client_id, local_epoch):
    return BATCH_SEED + round_id * 1_000_000 + client_id * 1_000 + local_epoch

def select_clients(round_id):
    if CLIENTS_PER_ROUND == NUM_CLIENTS:
        return list(range(NUM_CLIENTS))
    generator = torch.Generator(device='cpu').manual_seed(SELECTION_SEED + round_id)
    return torch.randperm(NUM_CLIENTS, generator=generator)[:CLIENTS_PER_ROUND].tolist()

def local_train(model, global_state, client_id, round_id):
    model.load_state_dict(global_state)
    model.train()
    optimizer = torch.optim.SGD(model.parameters(), lr=LEARNING_RATE)
    indices = client_indices[client_id]
    for local_epoch in range(LOCAL_EPOCHS):
        generator = torch.Generator(device='cpu').manual_seed(batch_seed(round_id, client_id, local_epoch))
        permutation = indices[torch.randperm(len(indices), generator=generator)].to(DEVICE)
        for start in range(0, len(permutation), BATCH_SIZE):
            batch_indices = permutation[start:start + BATCH_SIZE]
            loss = F.cross_entropy(model(train_features[batch_indices]), train_labels_gpu[batch_indices])
            optimizer.zero_grad(set_to_none=True)
            loss.backward()
            optimizer.step()
    return {name: parameter.detach().clone() for name, parameter in model.named_parameters()}, len(indices)

@torch.no_grad()
def flatten_state(state, names):
    return torch.cat([state[name].reshape(-1) for name in names])

def state_from_vector(vector, template, names):
    result, offset = {}, 0
    for name in names:
        size = template[name].numel()
        result[name] = vector[offset:offset + size].reshape(template[name].shape)
        offset += size
    return result

def spherical_kmeans(normalized_vectors, n_clusters, random_state, n_init=10, max_iter=100):
    # Input đã được chuẩn hóa một lần ở cấp round; tối ưu cosine similarity.
    rng = np.random.default_rng(random_state)
    best_labels, best_inertia = None, np.inf

    for _ in range(n_init):
        centers = normalized_vectors[rng.choice(len(normalized_vectors), n_clusters, replace=False)].copy()
        previous_labels = None
        for _ in range(max_iter):
            similarities = normalized_vectors @ centers.T
            labels = similarities.argmax(axis=1)
            if previous_labels is not None and np.array_equal(labels, previous_labels):
                break
            previous_labels = labels.copy()

            new_centers = np.zeros_like(centers)
            for cluster_id in range(n_clusters):
                members = normalized_vectors[labels == cluster_id]
                if len(members) == 0:
                    new_centers[cluster_id] = normalized_vectors[rng.integers(len(normalized_vectors))]
                else:
                    center = members.mean(axis=0)
                    new_centers[cluster_id] = center / max(np.linalg.norm(center), 1e-12)
            centers = new_centers

        similarities = normalized_vectors @ centers.T
        labels = similarities.argmax(axis=1)
        inertia = float(np.sum(1.0 - similarities[np.arange(len(labels)), labels]))
        if inertia < best_inertia:
            best_labels, best_inertia = labels.copy(), inertia

    return best_labels, best_inertia

def choose_elbow_k(normalized_vectors, max_k, round_id, fixed_k=None):
    max_k = min(max_k, len(normalized_vectors))
    if fixed_k is not None:
        selected_k = min(max(int(fixed_k), 1), max_k)
        labels, inertia = spherical_kmeans(
            normalized_vectors, selected_k, random_state=CLUSTER_SEED + round_id
        )
        return selected_k, np.array([selected_k]), np.array([inertia]), labels
    if max_k < 2:
        return 1, np.array([1]), np.array([0.0]), np.zeros(len(normalized_vectors), dtype=np.int64)
    ks = np.arange(2, max_k + 1)
    inertias, labels_by_k = [], {}
    for k in ks:
        labels, inertia = spherical_kmeans(
            normalized_vectors, int(k), random_state=CLUSTER_SEED + round_id
        )
        inertias.append(inertia)
        labels_by_k[int(k)] = labels
    points = np.column_stack([ks.astype(float), np.asarray(inertias)])
    normalized = (points - points.min(axis=0)) / np.maximum(points.max(axis=0) - points.min(axis=0), 1e-12)
    line = normalized[-1] - normalized[0]
    distances = np.abs(line[0] * (normalized[:, 1] - normalized[0, 1]) - line[1] * (normalized[:, 0] - normalized[0, 0])) / max(np.linalg.norm(line), 1e-12)
    selected_k = int(ks[int(np.argmax(distances))])
    return selected_k, ks, np.asarray(inertias), labels_by_k[selected_k]

def update_coassociation(co_sum, co_count, selected_clients, cluster_labels):
    client_ids = np.asarray(selected_clients, dtype=np.int64)
    labels = np.asarray(cluster_labels)
    for left in range(len(client_ids)):
        for right in range(left, len(client_ids)):
            client_left, client_right = client_ids[left], client_ids[right]
            co_count[client_left, client_right] += 1
            co_count[client_right, client_left] += 1
            same_cluster = float(labels[left] == labels[right])
            co_sum[client_left, client_right] += same_cluster
            co_sum[client_right, client_left] += same_cluster

def select_consensus_k(elbow_k_history, max_k):
    if not elbow_k_history:
        raise RuntimeError('Không có K trong warm-up để chọn K_final.')
    counts = Counter(int(k) for k in elbow_k_history)
    max_frequency = max(counts.values())
    tied_k = {k for k, frequency in counts.items() if frequency == max_frequency}
    # Nếu hòa, duyệt ngược lịch sử để chọn K xuất hiện ở round cuối hơn.
    for k in reversed(elbow_k_history):
        if int(k) in tied_k:
            return min(max(int(k), 1), max_k)
    raise RuntimeError('Không thể xác định K_final từ lịch sử warm-up.')


def build_consensus_client_clusters(co_sum, co_count, target_k):
    affinity = np.divide(
        co_sum, co_count, out=np.zeros_like(co_sum, dtype=np.float64), where=co_count > 0
    )
    np.fill_diagonal(affinity, 1.0)
    target_k = min(max(int(target_k), 1), len(affinity))
    clusters = [[client_id] for client_id in range(len(affinity))]
    while len(clusters) > target_k:
        best_pair, best_score = None, -np.inf
        for left in range(len(clusters)):
            for right in range(left + 1, len(clusters)):
                score = float(np.mean(affinity[np.ix_(clusters[left], clusters[right])]))
                if score > best_score:
                    best_pair, best_score = (left, right), score
        if best_pair is None:
            break
        left, right = best_pair
        clusters[left].extend(clusters[right])
        del clusters[right]
    mapping = np.full(len(affinity), -1, dtype=np.int64)
    for cluster_id, members in enumerate(clusters):
        mapping[members] = cluster_id
    return mapping, affinity

@torch.no_grad()
def get_cluster_feature_names(model_name, names):
    if model_name == 'cnn':
        cluster_names = [name for name in names if name.startswith('classifier.3.')]
        if not cluster_names:
            raise ValueError('Không tìm thấy lớp phân loại cuối (classifier.3) trong SmallCNN.')
        return cluster_names
    return list(names)

@torch.no_grad()
def top1_centroid_cosine_aggregate(client_states, client_sizes, global_state, names, round_id, client_cluster_mapping=None):
    global_vector = flatten_state(global_state, names)
    update_vectors = torch.stack([flatten_state(state, names) - global_vector for state in client_states])
    cluster_names = get_cluster_feature_names(MODEL_NAME, names)
    cluster_global_vector = flatten_state(global_state, cluster_names)
    cluster_update_vectors = torch.stack([flatten_state(state, cluster_names) - cluster_global_vector for state in client_states])
    cluster_update_norms = torch.linalg.vector_norm(cluster_update_vectors, dim=1, keepdim=True)
    normalized_cluster_vectors = cluster_update_vectors / cluster_update_norms.clamp_min(1e-12)
    normalized_updates = normalized_cluster_vectors.cpu().numpy()
    if client_cluster_mapping is None:
        selected_k, ks, inertias, cluster_labels = choose_elbow_k(
            normalized_updates, K_MAX, round_id
        )
    else:
        raw_labels = np.asarray(client_cluster_mapping)[np.asarray(selected_clients)]
        unique_labels = sorted(np.unique(raw_labels).tolist())
        remap = {old: new for new, old in enumerate(unique_labels)}
        cluster_labels = np.asarray([remap[label] for label in raw_labels], dtype=np.int64)
        selected_k = len(unique_labels)
        ks = np.asarray([selected_k])
        inertias = np.asarray([])
    server_aggregation = str(SERVER_AGGREGATION).strip().lower()
    if server_aggregation not in {'none', 'client', 'cluster'}:
        raise ValueError("SERVER_AGGREGATION phải là 'None', 'client' hoặc 'cluster'.")
    representative_indices = []
    cluster_sample_sizes = []
    for cluster_id in range(selected_k):
        members = np.flatnonzero(cluster_labels == cluster_id)
        member_indices = torch.as_tensor(members, device=update_vectors.device, dtype=torch.long)
        member_updates = normalized_cluster_vectors[member_indices]
        centroid = member_updates.mean(dim=0)
        centroid = centroid / torch.linalg.vector_norm(centroid).clamp_min(1e-12)
        cosine_similarities = member_updates @ centroid
        representative_indices.append(int(members[int(torch.argmax(cosine_similarities).item())]))
        cluster_sample_sizes.append(sum(client_sizes[i] for i in members))
    selected_updates = [update_vectors[i] for i in representative_indices]
    if server_aggregation == 'none':
        selected_sizes = torch.ones(
            len(representative_indices), dtype=update_vectors.dtype, device=update_vectors.device
        )
    elif server_aggregation == 'client':
        selected_sizes = torch.tensor(
            [client_sizes[i] for i in representative_indices],
            dtype=update_vectors.dtype,
            device=update_vectors.device,
        )
    else:  # cluster
        selected_sizes = torch.tensor(
            cluster_sample_sizes, dtype=update_vectors.dtype, device=update_vectors.device
        )
    selected_weights = selected_sizes / selected_sizes.sum()
    aggregated_update = (
        torch.stack(selected_updates) * selected_weights.unsqueeze(1)
    ).sum(dim=0)
    server_vector = global_vector + aggregated_update
    aggregated = state_from_vector(server_vector, global_state, names)
    return aggregated, selected_k, ks, inertias, cluster_labels, representative_indices

@torch.no_grad()
def evaluate(model, features, labels):
    model.eval()
    total_loss, total_correct = 0.0, 0
    for start in range(0, len(labels), EVAL_BATCH_SIZE):
        logits = model(features[start:start + EVAL_BATCH_SIZE])
        batch_labels = labels[start:start + EVAL_BATCH_SIZE]
        total_loss += F.cross_entropy(logits, batch_labels, reduction='sum').item()
        total_correct += (logits.argmax(dim=1) == batch_labels).sum().item()
    return total_loss / len(labels), total_correct / len(labels)
```

```python
METRIC_FIELDS = ['round', 'global_train_loss', 'global_train_accuracy', 'test_loss', 'test_accuracy', 'learning_rate', 'num_clients', 'num_selected_clients', 'selected_clients', 'num_samples', 'selected_k']

def load_metrics():
    if not METRICS_PATH.exists():
        return []
    with METRICS_PATH.open(newline='') as handle:
        return list(csv.DictReader(handle))

def truncate_metrics(round_id):
    rows = [row for row in load_metrics() if int(row['round']) <= round_id]
    tmp_path = METRICS_PATH.with_suffix('.tmp')
    with tmp_path.open('w', newline='') as handle:
        writer = csv.DictWriter(handle, fieldnames=METRIC_FIELDS)
        writer.writeheader()
        writer.writerows([{name: row.get(name, '') for name in METRIC_FIELDS} for row in rows])
    os.replace(tmp_path, METRICS_PATH)

def append_metric(row):
    if METRICS_PATH.exists():
        with METRICS_PATH.open(newline='') as handle:
            reader = csv.DictReader(handle)
            old_rows, old_fields = list(reader), reader.fieldnames or []
        if old_fields != METRIC_FIELDS:
            tmp_path = METRICS_PATH.with_suffix('.schema.tmp')
            with tmp_path.open('w', newline='') as handle:
                writer = csv.DictWriter(handle, fieldnames=METRIC_FIELDS)
                writer.writeheader()
                writer.writerows([{name: old_row.get(name, '') for name in METRIC_FIELDS} for old_row in old_rows])
            os.replace(tmp_path, METRICS_PATH)
    exists = METRICS_PATH.exists()
    with METRICS_PATH.open('a', newline='') as handle:
        writer = csv.DictWriter(handle, fieldnames=METRIC_FIELDS)
        if not exists:
            writer.writeheader()
        writer.writerow(row)
        handle.flush()
        os.fsync(handle.fileno())

def save_checkpoint(round_id, model, selected_clients):
    checkpoint = {'round': round_id, 'model_state': {name: value.detach().cpu() for name, value in model.named_parameters()}, 'dirichlet_alpha': DIRICHLET_ALPHA, 'partition_type': PARTITION_TYPE, 'partition_seed': PARTITION_SEED, 'model_seed': MODEL_SEED, 'batch_seed': BATCH_SEED, 'selection_seed': SELECTION_SEED, 'cluster_seed': CLUSTER_SEED, 'k_max': K_MAX, 'elbow_rounds': ELBOW_ROUNDS, 'representative_selection': 'closest_to_cluster_centroid_cosine', 'aggregation_vector': 'update_vector', 'server_aggregation': SERVER_AGGREGATION, 'clustering_algorithm': 'spherical', 'clustering_feature': 'last_classifier_layer', 'elbow_k_history': elbow_k_history, 'coassociation_sum': coassociation_sum, 'coassociation_count': coassociation_count, 'consensus_client_clusters': consensus_client_clusters, 'consensus_k': consensus_k, 'consensus_method': CONSENSUS_METHOD, 'selected_clients': selected_clients, 'torch_rng_state': torch.get_rng_state(), 'cuda_rng_state': torch.cuda.get_rng_state_all()}
    tmp_path = CHECKPOINT_PATH.with_suffix('.pt.tmp')
    torch.save(checkpoint, tmp_path)
    os.replace(tmp_path, CHECKPOINT_PATH)

current_config = {'model_name': MODEL_NAME, 'dataset_name': DATASET_NAME, 'imbalance_mode': IMBALANCE_MODE, 'num_clients': NUM_CLIENTS, 'clients_per_round': CLIENTS_PER_ROUND, 'local_epochs': LOCAL_EPOCHS, 'batch_size': BATCH_SIZE, 'learning_rate': LEARNING_RATE, 'resnet_size': RESNET_SIZE, 'dirichlet_alpha': DIRICHLET_ALPHA, 'partition_type': PARTITION_TYPE, 'partition_seed': PARTITION_SEED, 'model_seed': MODEL_SEED, 'batch_seed': BATCH_SEED, 'selection_seed': SELECTION_SEED, 'cluster_seed': CLUSTER_SEED, 'k_max': K_MAX, 'elbow_rounds': ELBOW_ROUNDS, 'consensus_method': CONSENSUS_METHOD, 'clustering_algorithm': 'spherical', 'clustering_feature': 'last_classifier_layer', 'representative_selection': 'closest_to_cluster_centroid_cosine', 'aggregation_vector': 'update_vector', 'server_aggregation': SERVER_AGGREGATION}
if not isinstance(ELBOW_ROUNDS, int) or not 1 <= ELBOW_ROUNDS <= ROUNDS:
    raise ValueError('ELBOW_ROUNDS phải là số nguyên trong khoảng từ 1 đến ROUNDS.')

elbow_k_history = []
coassociation_sum = np.zeros((NUM_CLIENTS, NUM_CLIENTS), dtype=np.float64)
coassociation_count = np.zeros((NUM_CLIENTS, NUM_CLIENTS), dtype=np.int64)
consensus_client_clusters = None
consensus_k = None
seed_everything(MODEL_SEED)
global_model = build_model()
start_round = 1
if CHECKPOINT_PATH.exists():
    if CONFIG_PATH.exists():
        with CONFIG_PATH.open() as handle:
            saved_config = json.load(handle)
        mismatches = [f'  - {key}: saved={saved_config.get(key, "<missing>")}, current={value}' for key, value in current_config.items() if key not in saved_config or saved_config[key] != value]
        if mismatches:
            raise ValueError(f'Run {RUN_ID} configuration changed:\n' + '\n'.join(mismatches) + '\nUse a new RUN_ID.')
        saved_config['rounds'] = ROUNDS
        with CONFIG_PATH.open('w') as handle:
            json.dump(saved_config, handle, indent=2)
    checkpoint = torch.load(CHECKPOINT_PATH, map_location='cpu')
    elbow_k_history = [int(k) for k in checkpoint.get('elbow_k_history', [])]
    coassociation_sum = np.asarray(checkpoint.get('coassociation_sum', coassociation_sum))
    coassociation_count = np.asarray(checkpoint.get('coassociation_count', coassociation_count))
    saved_mapping = checkpoint.get('consensus_client_clusters')
    saved_method = checkpoint.get('consensus_method')
    consensus_k = checkpoint.get('consensus_k')
    if saved_method == CONSENSUS_METHOD and saved_mapping is not None and consensus_k is not None:
        consensus_client_clusters = np.asarray(saved_mapping, dtype=np.int64)
        consensus_k = int(consensus_k)
    else:
        # Checkpoint cũ: dựng lại mapping từ ma trận và lịch sử K.
        consensus_client_clusters = None
        consensus_k = None
    global_model.load_state_dict(checkpoint['model_state'], strict=True)
    torch.set_rng_state(checkpoint['torch_rng_state'])
    torch.cuda.set_rng_state_all(checkpoint['cuda_rng_state'])
    completed_round = checkpoint['round']
    truncate_metrics(completed_round)
    start_round = completed_round + 1
    print(f'Resuming Run {RUN_ID} from round {start_round} to {ROUNDS}.')
else:
    with CONFIG_PATH.open('w') as handle:
        json.dump({'run_id': RUN_ID, 'run_dir': str(RUN_DIR), 'rounds': ROUNDS, **current_config, 'partition_path': str(partition_path), 'device': str(DEVICE)}, handle, indent=2)
    print(f'Starting Run {RUN_ID} from round 1 to {ROUNDS}.')

# Fallback cho checkpoint cũ chưa lưu lịch sử K: đọc lại từ metrics.csv.
if not elbow_k_history and METRICS_PATH.exists():
    elbow_k_history = [
        int(row['selected_k'])
        for row in load_metrics()
        if int(row['round']) <= ELBOW_ROUNDS and row.get('selected_k')
    ]
if start_round > ELBOW_ROUNDS and consensus_client_clusters is None:
    if not np.any(coassociation_count):
        raise RuntimeError('Không có ma trận clustering để tạo consensus mapping sau ELBOW_ROUNDS.')
    consensus_k = select_consensus_k(elbow_k_history, K_MAX)
    consensus_client_clusters, _ = build_consensus_client_clusters(
        coassociation_sum, coassociation_count, consensus_k
    )
    print(f'Consensus mapping loaded: K_final={consensus_k}, {len(np.unique(consensus_client_clusters))} clusters.')

if CLUSTER_LOG_PATH.exists():
    with CLUSTER_LOG_PATH.open('r') as handle:
        cluster_log = json.load(handle)
else:
    cluster_log = {}

worker_model = build_model()
names = [name for name, _ in global_model.named_parameters()]
for round_id in range(start_round, ROUNDS + 1):
    round_start = time.time()
    selected_clients = select_clients(round_id)
    global_state = {name: parameter.detach().clone() for name, parameter in global_model.named_parameters()}
    client_states, client_sizes = [], []
    for client_id in selected_clients:
        state, size = local_train(worker_model, global_state, client_id, round_id)
        client_states.append(state)
        client_sizes.append(size)
    if round_id > ELBOW_ROUNDS and consensus_client_clusters is None:
        consensus_k = select_consensus_k(elbow_k_history, K_MAX)
        consensus_client_clusters, _ = build_consensus_client_clusters(
            coassociation_sum, coassociation_count, consensus_k
        )
    round_mapping = None if round_id <= ELBOW_ROUNDS else consensus_client_clusters
    new_state, selected_k, ks, inertias, cluster_labels, representative_indices = top1_centroid_cosine_aggregate(
        client_states, client_sizes, global_state, names, round_id,
        client_cluster_mapping=round_mapping
    )
    if round_id <= ELBOW_ROUNDS:
        elbow_k_history.append(selected_k)
        update_coassociation(coassociation_sum, coassociation_count, selected_clients, cluster_labels)
        if round_id == ELBOW_ROUNDS:
            consensus_k = select_consensus_k(elbow_k_history, K_MAX)
            consensus_client_clusters, _ = build_consensus_client_clusters(
                coassociation_sum, coassociation_count, consensus_k
            )
            print(f'Consensus mapping created: K_final={consensus_k}, {len(np.unique(consensus_client_clusters))} clusters.')
    representative_clients = [selected_clients[i] for i in representative_indices]
    cluster_log[str(round_id)] = {
        'client_clusters': {
            str(client_id): int(label) + 1
            for client_id, label in zip(selected_clients, cluster_labels)
        },
        'representative_clients': {
            str(cluster_id + 1): client_id
            for cluster_id, client_id in enumerate(representative_clients)
        },
    }
    cluster_log_tmp_path = CLUSTER_LOG_PATH.with_suffix('.tmp')
    with cluster_log_tmp_path.open('w') as handle:
        json.dump(cluster_log, handle, indent=2, sort_keys=True)
    os.replace(cluster_log_tmp_path, CLUSTER_LOG_PATH)
    global_model.load_state_dict(new_state, strict=True)
    global_train_loss, global_train_accuracy = evaluate(global_model, train_features, train_labels_gpu)
    test_loss, test_accuracy = evaluate(global_model, test_features, test_labels_gpu)
    append_metric({'round': round_id, 'global_train_loss': global_train_loss, 'global_train_accuracy': global_train_accuracy, 'test_loss': test_loss, 'test_accuracy': test_accuracy, 'learning_rate': LEARNING_RATE, 'num_clients': NUM_CLIENTS, 'num_selected_clients': len(selected_clients), 'selected_clients': json.dumps(selected_clients), 'num_samples': sum(client_sizes), 'selected_k': selected_k})
    save_checkpoint(round_id, global_model, selected_clients)
    print(f'Round {round_id:02d}/{ROUNDS} | K={selected_k} | clients {selected_clients} | test acc {test_accuracy:.2%} | {time.time() - round_start:.1f}s')

del worker_model
torch.save({name: value.detach().cpu() for name, value in global_model.named_parameters()}, RUN_DIR / 'final_model_parameters.pt')
print('Finished. Results:', RUN_DIR)
```
